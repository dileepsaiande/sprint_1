```python
class InvalidFileException(Exception):
    """Raised when the provided file path is invalid."""
    pass


class UnsupportedFileFormatException(Exception):
    """Raised when the file format is not supported."""
    pass


print("Custom exceptions created successfully.")
```

```python
class DataLoader:
    """
    A reusable class for loading CSV, JSON and TXT files.
    """

    SUPPORTED_FORMATS = {".csv", ".json", ".txt"}

    def __init__(self):
        """Initialize the DataLoader."""
        pass

    @staticmethod
    def validate_file_path(file_path):
        """
        Validate that the given file path exists
        and points to a file.
        """

        path = Path(file_path)

        if not path.exists():
            raise InvalidFileException(
                f"File does not exist: {file_path}"
            )

        if not path.is_file():
            raise InvalidFileException(
                f"Path is not a file: {file_path}"
            )

        return path

    @classmethod
    def supported_formats(cls):
        """
        Return the supported file formats.
        """
        return cls.SUPPORTED_FORMATS

    def load(self, file_path):
        """
        Load data based on the file extension.
        """

        try:

            # Validate file path
            path = self.validate_file_path(file_path)

            # Get file extension
            extension = path.suffix.lower()

            # Check supported format
            if extension not in self.SUPPORTED_FORMATS:
                raise UnsupportedFileFormatException(
                    f"Unsupported file format: {extension}"
                )

            # Read CSV
            if extension == ".csv":
                return self._load_csv(path)

            # Read JSON
            elif extension == ".json":
                return self._load_json(path)

            # Read TXT
            elif extension == ".txt":
                return self._load_txt(path)

        except (
            InvalidFileException,
            UnsupportedFileFormatException
        ):
            raise

    @staticmethod
    def _load_csv(path):
        """
        Read CSV file and return a list of dictionaries.
        """

        with open(
            path,
            "r",
            encoding="utf-8",
            newline=""
        ) as file:

            reader = csv.DictReader(file)

            return list(reader)

    @staticmethod
    def _load_json(path):
        """
        Read JSON file and return Python data.
        """

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as file:

            return json.load(file)

    @staticmethod
    def _load_txt(path):
        """
        Read TXT file and return a list of lines.
        """

        with open(
            path,
            "r",
            encoding="utf-8"
        ) as file:

            return [
                line.strip()
                for line in file
            ]
```

```python
class DataWriter:
    """
    A reusable class for saving data into
    CSV, JSON, Pickle and TXT files.
    """

    def __init__(self):
        """Initialize the DataWriter."""
        pass

    @staticmethod
    def _create_parent_directory(file_path):
        """
        Create the parent directory if it does not exist.
        """

        path = Path(file_path)

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

    def save_csv(self, data, file_path):
        """
        Save a list of dictionaries into a CSV file.
        """

        self._create_parent_directory(file_path)

        if not data:
            raise ValueError(
                "Cannot save empty data to CSV."
            )

        with open(
            file_path,
            "w",
            encoding="utf-8",
            newline=""
        ) as file:

            writer = csv.DictWriter(
                file,
                fieldnames=data[0].keys()
            )

            writer.writeheader()
            writer.writerows(data)

        print(
            f"CSV file saved successfully: {file_path}"
        )

    def save_json(self, data, file_path):
        """
        Save Python data into a JSON file.
        """

        self._create_parent_directory(file_path)

        with open(
            file_path,
            "w",
            encoding="utf-8"
        ) as file:

            json.dump(
                data,
                file,
                indent=4
            )

        print(
            f"JSON file saved successfully: {file_path}"
        )

    def save_pickle(self, data, file_path):
        """
        Save Python data into a Pickle file.
        """

        self._create_parent_directory(file_path)

        with open(
            file_path,
            "wb"
        ) as file:

            pickle.dump(
                data,
                file
            )

        print(
            f"Pickle file saved successfully: {file_path}"
        )

    def save_txt(self, data, file_path):
        """
        Save data into a TXT file.
        Each item is written on a separate line.
        """

        self._create_parent_directory(file_path)

        with open(
            file_path,
            "w",
            encoding="utf-8"
        ) as file:

            for item in data:
                file.write(
                    f"{item}\n"
                )

        print(
            f"TXT file saved successfully: {file_path}"
        )
```

```python
def divide_values(first_value, second_value):
    """
    Divide two numbers and convert unexpected errors
    into a custom DataProcessingException.
    """

    try:

        if second_value == 0:
            raise DataProcessingException(
                "Division by zero is not allowed."
            )

        return first_value / second_value

    except DataProcessingException:
        raise

    except Exception as error:
        raise DataProcessingException(
            f"Data processing failed: {error}"
        ) from error
```

```python
from pathlib import Path
import csv
import logging
import sqlite3


class OOPException(Exception):
    """Base exception for the OOP assessment."""


class InvalidFileException(OOPException):
    """Raised when a file is invalid."""


class DatabaseConnectionException(OOPException):
    """Raised when database connection fails."""
```

```python
class DataLoader:
    """Loads employee data from CSV files."""

    SUPPORTED_FORMATS = {".csv"}

    def __init__(self, file_path: str):
        self._file_path = file_path

    @property
    def file_path(self) -> str:
        """Return the file path."""
        return self._file_path

    @staticmethod
    def validate_file(path: str) -> bool:
        """Check whether a file exists."""
        return Path(path).is_file()

    @classmethod
    def supported_formats(cls) -> set[str]:
        """Return supported file formats."""
        return cls.SUPPORTED_FORMATS

    def load(self) -> list[dict[str, str]]:
        """Load employee data from a CSV file."""
        if not self.validate_file(self._file_path):
            raise InvalidFileException(
                f"File not found: {self._file_path}"
            )

        with open(self._file_path, "r", encoding="utf-8") as file:
            return list(csv.DictReader(file))
```

```python
class DataWriter:
    """Writes employee data to CSV."""

    def __init__(self, file_path: str):
        self._file_path = file_path

    @property
    def file_path(self) -> str:
        """Return file path."""
        return self._file_path

    @staticmethod
    def validate(data: list[dict[str, str]]) -> bool:
        """Validate employee data."""
        return bool(data)

    def save(self, data: list[dict[str, str]]) -> None:
        """Save employee data."""
        if not self.validate(data):
            raise OOPException("Empty data cannot be saved.")

        with open(self._file_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=data[0].keys())
            writer.writeheader()
            writer.writerows(data)
```

```python
class LoggerManager:
    """Manages application logging."""

    def __init__(self, name: str = "EmployeeApp"):
        self._logger = logging.getLogger(name)
        self._logger.setLevel(logging.INFO)

    @property
    def logger(self) -> logging.Logger:
        """Return the logger."""
        return self._logger

    def log(self, message: str) -> None:
        """Log an information message."""
        self._logger.info(message)
```

```python
class DatabaseConnector:
    """Manages SQLite database connections."""

    def __init__(self, database: str = "employees.db"):
        self._database = database

    @property
    def database(self) -> str:
        """Return database name."""
        return self._database

    def connect(self) -> sqlite3.Connection:
        """Connect to the database."""
        try:
            return sqlite3.connect(self._database)
        except sqlite3.Error as e:
            raise DatabaseConnectionException(str(e))
```

```python
class ConfigManager:
    """Manages application configuration."""

    _config = {
        "app_name": "Employee Management System",
        "version": "1.0"
    }

    @classmethod
    def get(cls, key: str) -> str | None:
        """Get a configuration value."""
        return cls._config.get(key)

    @classmethod
    def set(cls, key: str, value: str) -> None:
        """Set a configuration value."""
        cls._config[key] = value

    @staticmethod
    def default_config() -> dict[str, str]:
        """Return default configuration."""
        return {"app_name": "Employee Management System"}
```

```python
import sqlite3

connection = sqlite3.connect("banking.db")
cursor = connection.cursor()

cursor.execute("PRAGMA foreign_keys = ON")

print("Banking database created successfully.")
```

```python
cursor.executescript("""
CREATE TABLE IF NOT EXISTS Customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL
);

CREATE TABLE IF NOT EXISTS Branches (
    branch_id INTEGER PRIMARY KEY,
    branch_name TEXT NOT NULL,
    city TEXT NOT NULL
);

CREATE TABLE IF NOT EXISTS Accounts (
    account_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    branch_id INTEGER NOT NULL,
    account_type TEXT NOT NULL,
    balance REAL DEFAULT 0,
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id),
    FOREIGN KEY (branch_id) REFERENCES Branches(branch_id)
);

CREATE TABLE IF NOT EXISTS Transactions (
    transaction_id INTEGER PRIMARY KEY,
    account_id INTEGER NOT NULL,
    amount REAL NOT NULL,
    transaction_type TEXT NOT NULL,
    FOREIGN KEY (account_id) REFERENCES Accounts(account_id)
);

CREATE TABLE IF NOT EXISTS Loans (
    loan_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    loan_amount REAL NOT NULL,
    loan_type TEXT NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id)
);
""")

connection.commit()

print("5 tables created successfully.")
```

```python
cursor.execute("""
SELECT
    name,
    balance,
    ROW_NUMBER() OVER (ORDER BY balance DESC) AS row_num
FROM Customers
JOIN Accounts USING (customer_id);
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
SELECT
    name,
    balance,
    RANK() OVER (ORDER BY balance DESC) AS rank,
    DENSE_RANK() OVER (ORDER BY balance DESC) AS dense_rank
FROM Customers
JOIN Accounts USING (customer_id);
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
SELECT
    name,
    balance,
    LAG(balance) OVER (ORDER BY balance) AS previous_balance,
    LEAD(balance) OVER (ORDER BY balance) AS next_balance
FROM Customers
JOIN Accounts USING (customer_id);
""")

for row in cursor.fetchall():
    print(row)
```
